```python
import pandas as pd
```

```python
df = pd.read_csv("src/credit_score_mlops/data/raw/creditcard.csv")

df = df.sample(frac=1,random_state=42).reset_index(drop=True)
split = int(0.7*len(df))
df_v1 = df.iloc[:split]
df_v2 = df.iloc[split:]

df_v1.to_csv("src/credit_score_mlops/data/raw/credit_card.csv", index=False)

df_v2.to_csv("src/credit_score_mlops/data/raw/credit_card_v2.csv",index=False)

print(f"v1 shape: {df_v1.shape}, v2 shape: {df_v2.shape}")
```
